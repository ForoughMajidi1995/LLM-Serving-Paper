```python
import pandas as pd
import ast

# Load file
df = df2

target_cols = [
    "merged_models",
    "model_family",
    "alignment_type",
    "modality",
    "scale",
    "architecture",
    "quantization",
    "context_length",
    "specialization",
    "deployment_context",
    "versioning",
]

def parse_to_list(x):
    # Case 1: already list → return as is
    if isinstance(x, list):
        return x

    # Case 2: numpy array → convert to list
    if hasattr(x, "tolist"):
        try:
            x = x.tolist()
            if isinstance(x, list):
                return x
        except Exception:
            pass

    # Case 3: scalar NaN / None
    try:
        if pd.isna(x):
            return []
    except Exception:
        pass

    # Case 4: string
    x = str(x).strip()
    if x == "" or x == "[]":
        return []

    try:
        parsed = ast.literal_eval(x)
        if isinstance(parsed, list):
            return parsed
        else:
            return [parsed]
    except Exception:
        return [x]
def merge_unique(series):
    seen = []
    for cell in series:
        values = parse_to_list(cell)
        for v in values:
            if pd.notna(v):
                v = str(v).strip()
                if v != "" and v not in seen:
                    seen.append(v)
    return seen

# Group by repo_full_name and collect unique merged values
new_df = (
    df.groupby("repo_full_name", as_index=False)[target_cols]
      .agg(merge_unique)
)

# Put repo_full_name first explicitly
new_df = new_df[["repo_full_name"] + target_cols]

# View result
new_df

# Optional: save it
```

```python
import pandas as pd
import ast
import numpy as np
from collections import Counter

df = non_empty_new_df

target_cols = [col for col in df.columns if col != "repo_full_name"]
print("Target columns for counting unique values: ", target_cols)
def parse_to_list(x):
    # None
    if x is None:
        return []

    # already a list
    if isinstance(x, list):
        return x

    # numpy array / pandas array
    if isinstance(x, (np.ndarray, pd.Series)):
        return [v for v in x.tolist() if pd.notna(v)]

    # scalar NaN
    try:
        if pd.isna(x):
            return []
    except Exception:
        pass

    # string cases
    s = str(x).strip()
    if s == "" or s == "[]":
        return []

    try:
        parsed = ast.literal_eval(s)
        if parsed is None:
            return []
        if isinstance(parsed, list):
            return parsed
        if isinstance(parsed, tuple):
            return list(parsed)
        if isinstance(parsed, (np.ndarray, pd.Series)):
            return list(parsed)
        return [parsed]
    except Exception:
        return [s]

value_counts_by_column = {}

for col in target_cols:
    counter = Counter()

    for cell in df[col]:
        values = parse_to_list(cell)

        # count each unique value once per row
        unique_values_in_row = {
            str(v).strip()
            for v in values
            if str(v).strip() != "" and str(v).strip().lower() != "nan"
        }

        counter.update(unique_values_in_row)

    result_df = pd.DataFrame({
        col: list(counter.keys()),
        "row_count": list(counter.values())
    }).sort_values("row_count", ascending=False).reset_index(drop=True)

    value_counts_by_column[col] = result_df
```
