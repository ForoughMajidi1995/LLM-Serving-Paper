```python
# for tool_name in ["sglang", "lmdeploy", "tensorrt_llm", "flashinfer"]:
for tool_name in ["vllm"]:

# for tool_name in ["sglang"]:
    repo_info_df = pd.read_csv(f'/Users/forough/VScodeProjects/VScode_serving/dataset/repo_info/New repo info/repos_{tool_name}_all_unique_detailed.csv')
    repo_info_df = repo_info_df.drop(columns=["Unnamed: 0"])
    # print(f'Processing tool: {tool_name}, total repos: {len(repo_info_df)}')
    removed_repos, cleaned_df = clean_repositories_df(repo_info_df)
    

    from pathlib import Path
    import re

    folder = Path(f"/Users/forough/VScodeProjects/VScode_serving/dataset/repo_info/old repo info/{tool_name}/")

    numbers = []

    for file in folder.glob(f"info_{tool_name}_chunk_*.csv"):
        print("inside loop:", file)

        match = re.search(r'(\d+)', file.stem)
        if match:
            numbers.append(int(match.group()))

    print(numbers)

    numbers = sorted(numbers)
    for n in numbers:
        repo_info = pd.read_csv(f'/Users/forough/VScodeProjects/VScode_serving/dataset/repo_info/old repo info/{tool_name}/info_{tool_name}_chunk_{n}.csv')
        print(f'Before cleaning, total code contents for {tool_name}: {len(repo_info)}')
        # repo_info_filtered = repo_info[~repo_info['repo_full_name'].isin(removed_repos)].reset_index(drop=True)
        print(repo_info.columns)
        repo_info_filtered = repo_info[~repo_info['repo_fullName'].isin(removed_repos)].reset_index(drop=True)
        print(f'After cleaning, total code content for {tool_name}: {len(repo_info_filtered)}')
        repo_info_filtered.rename(columns={"repo_fullName": "repo_full_name"}, inplace=True)
        repo_info_filtered.to_csv(f'/Users/forough/VScodeProjects/Vscode_serving/dataset/repo_info/old repo info/{tool_name}/cleaned_info_{tool_name}_chunk_{n}.csv', index=False)
        print("\n")
    
```
