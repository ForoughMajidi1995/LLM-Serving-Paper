```python
import pandas as pd

def count_unique_lists_sorted(df, column="appears_in"):
    """
    Returns a DataFrame of unique lists in `column`,
    sorted by descending occurrence count.
    """
    # keep only valid lists
    s = df[column].dropna().apply(
        lambda x: tuple(x) if isinstance(x, list) else None
    ).dropna()
    
    # count and sort
    counts = s.value_counts().reset_index()
    counts.columns = [column, "count"]
    
    # convert tuples back to lists
    counts[column] = counts[column].apply(list)
    
    return counts


result = count_unique_lists_sorted(A_updated)
print(result)
```

```python
import pandas as pd

# List of the 5 studied frameworks
framework_names = ["vllm", "sglang", "tensorrt_llm", "lmdeploy", "flashinfer"]

# Concatenate the 5 dataframes vertically
combined_df = pd.concat(
    [result_dfs[framework_name] for framework_name in framework_names],
    axis=0,
    ignore_index=True
)

tmp = combined_df.copy()
tmp["appears_in_key"] = tmp["appears_in"].apply(lambda x: tuple(sorted(x)))

result = (
    tmp.groupby("appears_in_key")["repo_full_name"]
       .nunique()
       .reset_index(name="n_unique_repo_full_name")
)

result["appears_in"] = result["appears_in_key"].apply(list)
result = result[["appears_in", "n_unique_repo_full_name"]]
# Keep only rows where appears_in has at least two items
result = result[result["appears_in"].apply(len) >= 2]

result = result[["appears_in", "n_unique_repo_full_name"]]


result = result.sort_values(
    by="n_unique_repo_full_name",
    ascending=False
)


print(result)
```
