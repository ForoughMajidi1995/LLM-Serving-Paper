```python
from pathlib import Path
import pandas as pd

# your list of tool names
tool_names = ["vllm","sglang", "lmdeploy", "tensorrt_llm", "flashinfer"]

# base folder containing one folder per tool
base_dir = Path("/Users/forough/VScodeProjects/Vscode_serving/dataset/repo_info/old repo info")

all_tools_dfs = []
per_tool_dfs = {}

for tool_name in tool_names:
    tool_folder = base_dir / tool_name

    # only files matching: cleaned_info_{tool_name}_chunk_{n}.csv
    csv_files = sorted(tool_folder.glob(f"cleaned_info_{tool_name}_chunk_*.csv"))

    if not csv_files:
        print(f"No matching files found for {tool_name}")
        continue

    dfs = []
    for file in csv_files:
        df = pd.read_csv(file)
        df["tool_name"] = tool_name
        dfs.append(df)

    # concatenate all matching files for this tool
    tool_df = pd.concat(dfs, ignore_index=True)

    per_tool_dfs[tool_name] = tool_df
    all_tools_dfs.append(tool_df)

# concatenate all tools together
if all_tools_dfs:
    final_df = pd.concat(all_tools_dfs, ignore_index=True)
else:
    final_df = pd.DataFrame()

# print(final_df.head())
# print(final_df.shape)
final_df = final_df[["repo_full_name", "tool_name"]]
print(len(final_df))
# final_df
```

```python
result_df = (
    final_df
    .dropna(subset=["repo_full_name", "tool_name"])
    .groupby("repo_full_name", as_index=False)["tool_name"]
    .agg(lambda x: sorted(x.unique().tolist()))
    .rename(columns={"tool_name": "Merge_Tools"})
)
print(result_df.columns)

result_df
```

```python
kept = kept.merge(
    result_df[["repo_full_name", "Merge_Tools"]],
    on="repo_full_name",
    how="left"
)

kept["Merge_Tools"] = kept["Merge_Tools"].apply(
    lambda x: x if isinstance(x, list) else []
)

print(kept.columns)

kept
```

```python
result_df = (
    kept
    .explode("Merge_Tools")
    .groupby(["Merge_Tools", f"{title}_{summary_level}_initial_k_detected_topic"])
    .size()
    .reset_index(name="count")
    .rename(columns={
        "Merge_Tools": "tool_name",
        f"{title}_{summary_level}_initial_k_detected_topic": "topic_number"
    })
    .sort_values(["tool_name", "topic_number"])
    .reset_index(drop=True)
)
result_df = result_df[result_df["topic_number"] != -1].reset_index(drop=True)
# print(result_df["count"].tolist())

# result_df
```

```python
tool_order = ["vllm", "sglang", "tensorrt_llm", "lmdeploy", "flashinfer"]

result_df["tool_name"] = pd.Categorical(
    result_df["tool_name"],
    categories=tool_order,
    ordered=True
)

result_df = (
    result_df
    .sort_values(["tool_name", "topic_number"])
    .reset_index(drop=True)
)
print(result_df["count"].tolist())
result_df
```
